```python
from pathlib import Path
import splitfolders
```

```python
# Root project
PROJECT_ROOT = Path(".")

# Raw dataset path
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

# Folder chứa dataset brain tumor
BRAIN_DATA_DIR = RAW_DATA_DIR / "brain_data"

# Folder output sau khi chia dữ liệu
PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"
```

```python
print(f"Raw data path      : {RAW_DATA_DIR}")
print(f"Brain dataset path : {BRAIN_DATA_DIR}")
print(f"Processed path     : {PROCESSED_DATA_DIR}")
```

```python
if not BRAIN_DATA_DIR.exists():
    raise FileNotFoundError(
        f"Dataset folder not found: {BRAIN_DATA_DIR}"
    )

print("\nDataset folder found!")
```

70% train
15% validation
15% test

```python
splitfolders.ratio(
    input=str(BRAIN_DATA_DIR),        # Folder dataset gốc
    output=str(PROCESSED_DATA_DIR),   # Folder output
    seed=42,                          # Random seed
    ratio=(0.7, 0.15, 0.15),          # Train/Val/Test
    group_prefix=None,
    move=False                        # False = copy dữ liệu
)

print("\nDataset splitting completed!")
```

```python
for split in ["train", "val", "test"]:

    split_path = PROCESSED_DATA_DIR / split

    print(f"\n{split.upper()} DATASET")

    for class_dir in sorted(split_path.iterdir()):

        if class_dir.is_dir():

            image_count = len(list(class_dir.glob("*")))

            print(f"{class_dir.name:<15} : {image_count} images")
```
